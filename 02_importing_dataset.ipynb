# Importing Data Sets

### Importing a CSV into Python

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

# use read_csv to read the daat from the URL and store it in a DataFram

df = pd.read_csv(url)

# read_csv assumes the data contains a header

# Importing a csv without a header 
# if you data has no header, we need to assign headers by setting header to None

df = pd.read_csv(url, header=None)
```

### Printing the dataframe

- **df** prints the entire dataframe (not recommended for larger datasets)
- **df.head(n)** to show the first n rows of data frame
- **df.tail(n)** to show the bottom n rows of data frame

```python
df.head(5)
```

### Adding headers
- Replace default headers (by df.columns = headers)


```python
headers = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l"]

df.columns = headers

df.head(5)
```

### Exporting a Pandas dataframe to CSV

### Data types

We use `dataframe.datatypes` to check data types
When the data type method is applied to the dataset, the data type of each column is returned in a series

```python
df.dtypes
```

### dataframe.describe()
- Returns a statistical summary of each column to learn about the distribution of data in each column. 
- The statistical metrics can tell the data scientist if there are mathematical issues that may exist like
   extreme outliers and large deviations
- Data frame descrive() function skips rows and columns that do not contains numbers, it is possible to make the described method  
  work for object type columns as well.

```python
df.describe()
```

### dataframe.describe(inclule="all")

- It will provide full summary statistics
- now it will show summary of all columns including object type attriburtes


```python
df.describe(include="all")
```
